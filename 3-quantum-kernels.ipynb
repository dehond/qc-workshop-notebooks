# Quantum kernels
To get a better idea of the capabilities of current neutral-atom quantum computers, we'll have a look at the recent application presented in this paper: [arxiv:2605.21276](https://arxiv.org/abs/2605.21276). Here, they tried solving a differential equations using parametrized quantum circuits. Refer to the preprint for the mathematical details, here we'll purely look at the circuit they used.

Also note that in the paper the circuit was also implemented using an error-detecting code. In such an implementation, each physical qubit is replaced by a logical one (which consists of multiple physical qubits). Using the redundancy in such a state, we can detect and correct errors that creep in during the circuit.
